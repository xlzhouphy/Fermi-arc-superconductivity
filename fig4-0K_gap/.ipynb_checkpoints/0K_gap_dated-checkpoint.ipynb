```python
#s-wave theta as parameter
import numpy as np
from scipy.optimize import root

t = 1.0
t1 = 0.0
t2 = 0.0
g = 2.0*t
mu = 0.0

N = 64
k_points = np.linspace(-np.pi, np.pi, N, endpoint=False)
kx, ky = np.meshgrid(k_points, k_points)
Total_K_Points = N**2

def gamma_func(kx, ky):
    gamma = 1.0
    #gamma = np.cos(kx) - np.cos(ky)
    return gamma

def dispersion(kx, ky):
    ek = -2 * t * (np.cos(kx) + np.cos(ky))
    ek -= 4 * t1 * np.cos(kx) * np.cos(ky)
    ek -= 2 * t2 * (np.cos(2*kx) + np.cos(2*ky))
    ek -= mu
    return ek

xi_2d = dispersion(kx, ky)
g_2d = gamma_func(kx, ky)

def bcs_2d_system(theta_flat):
    theta = theta_flat.reshape(N, N)
    sin_2t = np.sin(2 * theta)
    cos_2t = np.cos(2 * theta)
    residual = 2 * xi_2d * sin_2t - g_2d * cos_2t * (g / Total_K_Points) * np.sum(g_2d * sin_2t) 
    return residual.flatten()

delta_guess = 0.37
initial_theta = 0.5 * np.arctan2(delta_guess * g_2d, xi_2d)
initial_theta_flat = initial_theta.flatten()

sol = root(
    bcs_2d_system, 
    initial_theta_flat, 
    method = 'df-sane', 
    tol=1e-6,
)

if sol.success:
    theta_res = sol.x.reshape(N, N)
    delta_final = g/2/Total_K_Points * np.sum(g_2d * np.sin(2 * theta_res))
    print(f" Delta: {delta_final:.6f}")
else:
    print("error", sol.message)
```

```python
#Fermi arc model Tc equation 25 (non-interacting interacting splited)
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import fsolve
from scipy.optimize import brentq
from scipy.optimize import root
import time

kB = 1.0
t = 1.0
t1 = -0.2
t2 = 0.1
g = 2.0

Q = [np.pi, np.pi]

U = 2.0

Nk = 64
kx_list = np.linspace(-np.pi, np.pi, Nk, endpoint=False)
ky_list = np.linspace(-np.pi, np.pi, Nk, endpoint=False)
KX, KY = np.meshgrid(kx_list, ky_list)
N = Nk * Nk

k_points = np.linspace(-np.pi, np.pi, Nk, endpoint=False)
kx, ky = np.meshgrid(k_points, k_points)
Total_K_Points = Nk**2

def band(kx, ky):
    return -2.0*t*(np.cos(kx) + np.cos(ky)) - 4.0*t1*np.cos(kx)*np.cos(ky) - 2*t2*(np.cos(2*kx) + np.cos(2*ky))

def band_Q(kx, ky):
    return -2.0*t*(np.cos(kx+Q[0]) + np.cos(ky+Q[1])) - 4.0*t1*np.cos(kx+Q[0])*np.cos(ky+Q[1]) - 2*t2*(np.cos(2*(kx+Q[0])) + np.cos(2*(ky+Q[1])))

def occupation_Q_per_spin(epsilon, epsilon_Q, mu, U, T):
    epsilon_minus_mu = epsilon - mu
    epsilon_Q_minus_mu = epsilon_Q - mu
    if T < 1e-8:
        T = 1e-8
    beta = 1.0/T
    E_1 = -beta*epsilon_Q_minus_mu
    E_2 = -beta*(epsilon_minus_mu + epsilon_Q_minus_mu + U)
    E_3 = -beta*epsilon_minus_mu
    E_0 = np.full_like(E_1, 0)
    E_max = np.maximum.reduce([E_0, E_1, E_2, E_3])
    numerator = np.exp(E_1 - E_max) + np.exp(E_2 - E_max)
    denominator = np.exp(E_0 - E_max) + np.exp(E_3 - E_max) + np.exp(E_1 - E_max) + np.exp(E_2 - E_max)
    return numerator/denominator

def chi0_hk(T, mu, U):
    if T < 1e-8:
        T = 1e-8
    beta = 1.0/T
    epsilon = band(KX, KY)
    epsilon_Q = band_Q(KX, KY)
    n_k = occupation_Q_per_spin(epsilon, epsilon_Q, mu, U, T)
    epsilon_minus_mu = epsilon - mu
    epsilon_Q_minus_mu = epsilon_Q - mu
    gamma = np.cos(KX) - np.cos(KY)
    #gamma = 1
    
    term_1_denominator = 2.0*epsilon_minus_mu
    term_1_numerator = gamma*gamma*(1.0 - n_k)*np.tanh(beta*epsilon_minus_mu/2.0)
    term_1 = np.where(np.abs(epsilon_minus_mu) < 1e-6, 
                      (1.0 - n_k)*beta/4.0,
                     term_1_numerator/term_1_denominator)
    
    epsilon_plus_U = epsilon_minus_mu + U
    term_2_denominator = 2.0*epsilon_plus_U
    term_2_numerator = gamma*gamma*n_k*np.tanh(beta*epsilon_plus_U/2.0)
    term_2 = np.where(np.abs(epsilon_plus_U) < 1e-6, 
                      n_k*beta/4.0,
                     term_2_numerator/term_2_denominator)

    term_3_denominator = 2.0*epsilon_minus_mu + U
    term_3_numerator = gamma*gamma*n_k*(1-n_k)*U*(np.tanh(beta*epsilon_plus_U/2.0)/2/epsilon_plus_U - np.tanh(beta*epsilon_minus_mu/2.0)/2/epsilon_minus_mu)
    term_3 = term_3_numerator/term_3_denominator
    #term_3 = 0
    
    chi0 = np.sum(term_1 + term_2 + term_3)/N
    return chi0

def tc_equation(T, mu, U, g):
    return g*chi0_hk(T, mu, U) - 1.0

def find_Tc(mu, U, g):
    T_max = 1.0
    T_min = 1e-7
    if tc_equation(T_max, mu, U, g) > 0:
        T_high = T_max
        while tc_equation(T_high, mu, U, g) > 0 and T_high < 10.0:
            T_high *= 2
        T_max = T_high
        print('T_high=', T_max)
    if tc_equation(T_min, mu, U, g) < 0:
        print('mu =', mu, 'low Tc')
        return 0.0
    try:
        Tc = brentq(tc_equation, T_min, T_max, args=(mu, U, g))
        return Tc
    except ValueError:
        return 0.0
    return Tc

def average_occupation(mu, T, U, KX, KY, N):
    epsilon = band(KX, KY)
    epsilon_Q = band_Q(KX, KY)
    n_k = occupation_Q_per_spin(epsilon, epsilon_Q, mu, U, T)
    return 2.0*np.sum(n_k) / N

def gamma_func(kx, ky):
    #gamma = 1.0
    gamma = np.cos(kx) - np.cos(ky)
    return gamma

def dispersion(kx, ky):
    ek = -2 * t * (np.cos(kx) + np.cos(ky))
    ek -= 4 * t1 * np.cos(kx) * np.cos(ky)
    ek -= 2 * t2 * (np.cos(2*kx) + np.cos(2*ky))
    ek -= mu
    return ek

kmap = np.empty(Nk, dtype=np.int32)

for i in range(Nk):
    kmap[i] = (i+Nk//2)%Nk

def bcs_2d_system(theta_flat):
    theta = theta_flat.reshape(Nk, Nk)
    
    sin_2t = np.sin(2 * theta)
    cos_2t = np.cos(2 * theta)

    sin_q = np.sin(theta[np.ix_(kmap, kmap)])
    
    gap_sum = 0.5*g/Total_K_Points * np.sum(g_2d * sin_2t)
    
    residual = 2 * xi_2d * sin_2t + 2*U*sin_2t*sin_q*sin_q- 2 * g_2d * cos_2t * gap_sum
    return residual.flatten()

mu_list = np.linspace(0.70, -2.55, 150)  #U=2.0
#mu_list = np.linspace(0.5, -2.75, 150)  #U=1.7
#mu_list = np.linspace(0.1, -3.15, 150)  #U=1.4
#mu_list = np.linspace(-0.15, -3.4, 150)  #U=1.1

Tc_results = []
Doping_x_results = []
delta_results = []

start_time = time.time()
for mu in mu_list:
    Tc = find_Tc(mu, U, g)
    T_threshold = 1e-6
    if Tc > T_threshold:
        avg_n = average_occupation(mu, Tc, U, KX, KY, N)
        doping_x = 1.0 - avg_n
    else:
        avg_n = average_occupation(mu, T_threshold, U, KX, KY, N)
        doping_x = 1.0 - avg_n
    print('dopping_x =', doping_x, 'Tc =', Tc)
    Tc_results.append(Tc)
    Doping_x_results.append(doping_x)
    
    xi_2d = dispersion(kx, ky)
    g_2d = gamma_func(kx, ky)
    delta_guess = 0.2
    initial_theta = 0.5 * np.arctan2(delta_guess * g_2d, xi_2d)
    initial_theta_flat = initial_theta.flatten()
    delta = 0
    sol = root(
        bcs_2d_system, 
        initial_theta_flat, 
        method='df-sane',
        tol=1e-6,
    )
    if sol.success:
        theta_res = sol.x.reshape(Nk, Nk)
        delta = g/2/Total_K_Points * np.sum(g_2d * np.sin(2 * theta_res))
        print(f" Delta: {delta:.6f}")
    else:
        print("error", sol.message)
    delta_results.append(delta)

x_plot = np.array(Doping_x_results)
Tc_plot = np.array(Tc_results)
delta_plot = np.array(delta_results)

mask_lower = x_plot >= 0.1
mask_upper = x_plot < 0.8
mask_combined = mask_lower & mask_upper

Tc_final = Tc_plot[mask_combined]
x_final = x_plot[mask_combined]
delta_final = delta_plot[mask_combined]

sort_indices = np.argsort(x_final)
x_final = x_final[sort_indices]
Tc_final = Tc_final[sort_indices]

end_time = time.time()
print(end_time-start_time)

plt.figure(figsize=(8, 8))
#plt.plot(x_final, Tc_final, 'ro-', linewidth=2, markersize=4, label=f'$U={U}, g={g}$, interacting')
#plt.plot(x_final_0, Tc_final_0, 'co-', linewidth=2, markersize=4, label=f'$U={U}, g={g}, non-interacting$')
plt.plot(x_final, Tc_final, 'o-', color='tab:orange', linewidth=2, markersize=4, label=f'$U={U}, g={g}$, interacting')
plt.title(f'Critical Temperature $T_c$ vs Doping $x = 1 - \\langle n \\rangle$')
plt.xlabel('Doping $x = 1 - \\langle n \\rangle$')
plt.ylabel('Critical Temperature $T_c / t$')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlim(0, 0.8)
plt.ylim(bottom=0)
plt.legend()
plt.show() 
```

```python
#Fermi arc model Tc equation 25 (non-interacting interacting splited)
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import fsolve
from scipy.optimize import brentq
from scipy.optimize import root
import time

kB = 1.0
t = 1.0
t1 = -0.2
t2 = 0.1
g = 2.0

Q = [np.pi, np.pi]

U = 0.0

Nk = 64
kx_list = np.linspace(-np.pi, np.pi, Nk, endpoint=False)
ky_list = np.linspace(-np.pi, np.pi, Nk, endpoint=False)
KX, KY = np.meshgrid(kx_list, ky_list)
N = Nk * Nk

k_points = np.linspace(-np.pi, np.pi, Nk, endpoint=False)
kx, ky = np.meshgrid(k_points, k_points)
Total_K_Points = Nk**2

def band(kx, ky):
    return -2.0*t*(np.cos(kx) + np.cos(ky)) - 4.0*t1*np.cos(kx)*np.cos(ky) - 2*t2*(np.cos(2*kx) + np.cos(2*ky))

def band_Q(kx, ky):
    return -2.0*t*(np.cos(kx+Q[0]) + np.cos(ky+Q[1])) - 4.0*t1*np.cos(kx+Q[0])*np.cos(ky+Q[1]) - 2*t2*(np.cos(2*(kx+Q[0])) + np.cos(2*(ky+Q[1])))

def occupation_Q_per_spin(epsilon, epsilon_Q, mu, U, T):
    epsilon_minus_mu = epsilon - mu
    epsilon_Q_minus_mu = epsilon_Q - mu
    if T < 1e-8:
        T = 1e-8
    beta = 1.0/T
    E_1 = -beta*epsilon_Q_minus_mu
    E_2 = -beta*(epsilon_minus_mu + epsilon_Q_minus_mu + U)
    E_3 = -beta*epsilon_minus_mu
    E_0 = np.full_like(E_1, 0)
    E_max = np.maximum.reduce([E_0, E_1, E_2, E_3])
    numerator = np.exp(E_1 - E_max) + np.exp(E_2 - E_max)
    denominator = np.exp(E_0 - E_max) + np.exp(E_3 - E_max) + np.exp(E_1 - E_max) + np.exp(E_2 - E_max)
    return numerator/denominator

def chi0_hk(T, mu, U):
    if T < 1e-8:
        T = 1e-8
    beta = 1.0/T
    epsilon = band(KX, KY)
    epsilon_Q = band_Q(KX, KY)
    n_k = occupation_Q_per_spin(epsilon, epsilon_Q, mu, U, T)
    epsilon_minus_mu = epsilon - mu
    epsilon_Q_minus_mu = epsilon_Q - mu
    gamma = np.cos(KX) - np.cos(KY)
    #gamma = 1
    
    term_1_denominator = 2.0*epsilon_minus_mu
    term_1_numerator = gamma*gamma*(1.0 - n_k)*np.tanh(beta*epsilon_minus_mu/2.0)
    term_1 = np.where(np.abs(epsilon_minus_mu) < 1e-6, 
                      (1.0 - n_k)*beta/4.0,
                     term_1_numerator/term_1_denominator)
    
    epsilon_plus_U = epsilon_minus_mu + U
    term_2_denominator = 2.0*epsilon_plus_U
    term_2_numerator = gamma*gamma*n_k*np.tanh(beta*epsilon_plus_U/2.0)
    term_2 = np.where(np.abs(epsilon_plus_U) < 1e-6, 
                      n_k*beta/4.0,
                     term_2_numerator/term_2_denominator)

    term_3_denominator = 2.0*epsilon_minus_mu + U
    term_3_numerator = gamma*gamma*n_k*(1-n_k)*U*(np.tanh(beta*epsilon_plus_U/2.0)/2/epsilon_plus_U - np.tanh(beta*epsilon_minus_mu/2.0)/2/epsilon_minus_mu)
    term_3 = term_3_numerator/term_3_denominator
    #term_3 = 0
    
    chi0 = np.sum(term_1 + term_2 + term_3)/N
    return chi0

def tc_equation(T, mu, U, g):
    return g*chi0_hk(T, mu, U) - 1.0

def find_Tc(mu, U, g):
    T_max = 1.0
    T_min = 1e-7
    if tc_equation(T_max, mu, U, g) > 0:
        T_high = T_max
        while tc_equation(T_high, mu, U, g) > 0 and T_high < 10.0:
            T_high *= 2
        T_max = T_high
        print('T_high=', T_max)
    if tc_equation(T_min, mu, U, g) < 0:
        print('mu =', mu, 'low Tc')
        return 0.0
    try:
        Tc = brentq(tc_equation, T_min, T_max, args=(mu, U, g))
        return Tc
    except ValueError:
        return 0.0
    return Tc

def average_occupation(mu, T, U, KX, KY, N):
    epsilon = band(KX, KY)
    epsilon_Q = band_Q(KX, KY)
    n_k = occupation_Q_per_spin(epsilon, epsilon_Q, mu, U, T)
    return 2.0*np.sum(n_k) / N

def gamma_func(kx, ky):
    #gamma = 1.0
    gamma = np.cos(kx) - np.cos(ky)
    return gamma

def dispersion(kx, ky):
    ek = -2 * t * (np.cos(kx) + np.cos(ky))
    ek -= 4 * t1 * np.cos(kx) * np.cos(ky)
    ek -= 2 * t2 * (np.cos(2*kx) + np.cos(2*ky))
    ek -= mu
    return ek

kmap = np.empty(Nk, dtype=np.int32)

for i in range(Nk):
    kmap[i] = (i+Nk//2)%Nk

def bcs_2d_system(theta_flat):
    theta = theta_flat.reshape(Nk, Nk)
    
    sin_2t = np.sin(2 * theta)
    cos_2t = np.cos(2 * theta)

    sin_q = np.sin(theta[np.ix_(kmap, kmap)])
    
    gap_sum = 0.5*g/Total_K_Points * np.sum(g_2d * sin_2t)
    
    residual = 2 * xi_2d * sin_2t + 2*U*sin_2t*sin_q*sin_q- 2 * g_2d * cos_2t * gap_sum
    return residual.flatten()

mu_list = np.linspace(0.70, -2.55, 150)  #U=2.0
#mu_list = np.linspace(0.5, -2.75, 150)  #U=1.7
#mu_list = np.linspace(0.1, -3.15, 150)  #U=1.4
#mu_list = np.linspace(-0.15, -3.4, 150)  #U=1.1

Tc_results_0 = []
Doping_x_results_0 = []
delta_results_0 = []

start_time = time.time()
for mu in mu_list:
    Tc = find_Tc(mu, U, g)
    T_threshold = 1e-6
    if Tc > T_threshold:
        avg_n = average_occupation(mu, Tc, U, KX, KY, N)
        doping_x = 1.0 - avg_n
    else:
        avg_n = average_occupation(mu, T_threshold, U, KX, KY, N)
        doping_x = 1.0 - avg_n
    print('dopping_x =', doping_x, 'Tc =', Tc)
    Tc_results_0.append(Tc)
    Doping_x_results_0.append(doping_x)
    
    xi_2d = dispersion(kx, ky)
    g_2d = gamma_func(kx, ky)
    delta_guess = 0.2
    initial_theta = 0.5 * np.arctan2(delta_guess * g_2d, xi_2d)
    initial_theta_flat = initial_theta.flatten()
    delta_0 = 0
    sol = root(
        bcs_2d_system, 
        initial_theta_flat, 
        method='df-sane',
        tol=1e-6,
    )
    if sol.success:
        theta_res = sol.x.reshape(Nk, Nk)
        delta_0 = g/2/Total_K_Points * np.sum(g_2d * np.sin(2 * theta_res))
        print(f" Delta: {delta_0:.6f}")
    else:
        print("error", sol.message)
    delta_results_0.append(delta_0)

x_plot_0 = np.array(Doping_x_results_0)
Tc_plot_0 = np.array(Tc_results_0)
delta_plot_0 = np.array(delta_results_0)

mask_lower = x_plot_0 >= 0.1
mask_upper = x_plot_0 < 0.8
mask_combined = mask_lower & mask_upper

Tc_final_0 = Tc_plot_0[mask_combined]
x_final_0 = x_plot_0[mask_combined]
delta_final_0 = delta_plot_0[mask_combined]

sort_indices = np.argsort(x_final_0)
x_final_0 = x_final_0[sort_indices]
Tc_final_0 = Tc_final_0[sort_indices]

end_time = time.time()
print(end_time-start_time)

plt.figure(figsize=(8, 8))
#plt.plot(x_final, Tc_final, 'ro-', linewidth=2, markersize=4, label=f'$U={U}, g={g}$, interacting')
#plt.plot(x_final_0, Tc_final_0, 'co-', linewidth=2, markersize=4, label=f'$U={U}, g={g}, non-interacting$')
plt.plot(x_final_0, Tc_final_0, 'o-', color='tab:orange', linewidth=2, markersize=4, label=f'$U={U}, g={g}$, interacting')
plt.title(f'Critical Temperature $T_c$ vs Doping $x = 1 - \\langle n \\rangle$')
plt.xlabel('Doping $x = 1 - \\langle n \\rangle$')
plt.ylabel('Critical Temperature $T_c / t$')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlim(0, 0.8)
plt.ylim(bottom=0)
plt.legend()
plt.show() 
```
