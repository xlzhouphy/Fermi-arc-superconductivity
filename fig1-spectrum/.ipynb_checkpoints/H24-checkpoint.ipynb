```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("publication.mplstyle")

def hamiltonian(k, t0):
    kx, ky = k
    # 定义哈密顿量矩阵
    H = np.array([
        [0, t0, np.exp(1j*(kx + ky)), t0],
        [t0, 0, t0, np.exp(1j*(ky - kx))],
        [np.exp(-1j*(kx + ky)), t0, 0, t0],
        [t0, np.exp(1j*(kx - ky)), t0, 0]
    ], dtype=complex)
    return H

# 1. 定义高对称点 (单位: pi)
G = np.array([0, 0])
X = np.array([np.pi, 0])
M = np.array([np.pi, np.pi])
Y = np.array([0, np.pi])

# 2. 设定路径和采样点数
points = [G, X, M, Y, G]
labels = [r'$\Gamma$', 'X', 'M', 'Y', r'$\Gamma$']
n_samples = 100  # 每段路径的采样点数

k_path = []
k_ticks = [0]
current_dist = 0

for i in range(len(points) - 1):
    start_pt = points[i]
    end_pt = points[i+1]
    # 在两点之间插值
    for t in np.linspace(0, 1, n_samples, endpoint=False):
        k_path.append(start_pt + t * (end_pt - start_pt))
    
    # 计算路径长度用于绘图
    dist = np.linalg.norm(end_pt - start_pt)
    current_dist += dist
    k_ticks.append(current_dist)

k_path = np.array(k_path)
distances = np.linspace(0, k_ticks[-1], len(k_path))

# 3. 计算特征值 (能带)
bands = []
t0_val = 0.4 # 你可以根据需要调整 t0 的值

for k in k_path:
    H = hamiltonian(k, t0=t0_val)
    eigenvalues = np.linalg.eigvalsh(H) # 使用 eigvalsh 针对 Hermite 矩阵更快更稳定
    bands.append(eigenvalues)

bands = np.array(bands)

# 4. 绘图
for i in range(bands.shape[1]):
    plt.plot(distances, bands[:, i], color = 'blue',label=f'Band {i+1}' if i==0 else "")

# 装饰图形
plt.xticks(k_ticks, labels)
for tick in k_ticks:
    plt.axvline(tick, color='gray', linestyle='--', alpha=0.5)

plt.xlim([0, distances[-1]])
plt.title(f'$H_2^{(4)}$ Band Structure ($t_0 = {t0_val}$)')
plt.xlabel('k-path')
plt.ylabel('Energy E')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 参数设置 ---
m = 20
gamma = 0.4

# 矩阵维度为 4*m*m
H = np.zeros((4*m*m, 4*m*m))

# --- 构造哈密顿量 (严格遵循原逻辑) ---
for j in range(1, m + 1):
    for i in range(1, m + 1):
        idx0 = (j-1)*4*m + (i-1)*4 + 0
        idx1 = (j-1)*4*m + (i-1)*4 + 1
        idx2 = (j-1)*4*m + (i-1)*4 + 2
        idx3 = (j-1)*4*m + (i-1)*4 + 3

        H[idx0, idx1] = gamma
        H[idx1, idx0] = gamma

        H[idx1, idx2] = gamma
        H[idx2, idx1] = gamma
    
        H[idx2, idx3] = gamma
        H[idx3, idx2] = gamma

        H[idx3, idx0] = gamma
        H[idx0, idx3] = gamma

        if i != m  and j != m :
            H[idx0, (j)*4*m + (i)*4 + 2] = 1
            H[(j)*4*m + (i)*4 + 2, idx0] = 1

        if i != 1 and j != 1:
            H[idx2, (j-2)*4*m + (i-2)*4 + 0] = 1
            H[(j-2)*4*m + (i-2)*4 + 0, idx2] = 1

        if i != m  and j != 1 :
            H[idx3, (j-2)*4*m + (i)*4 + 1] = 1
            H[(j-2)*4*m + (i)*4 + 1, idx3] = 1

        if i != 1  and j != m :
            H[idx1, (j)*4*m + (i-2)*4 + 3] = 1
            H[(j)*4*m + (i-2)*4 + 3, idx1] = 1

# --- 对角化与排序 ---
e1, V = np.linalg.eigh(H) # eigh 自动排序并保证实数
# 虽然 eigh 已排序，但为了严格对应 Matlab 的 sort：
e1 = np.sort(e1)

# 对应 Matlab 的 V = round(V, 2)
V = np.round(V, 2)

# --- 绘图 ---
n = np.arange(1, len(e1) + 1)
plt.figure(figsize=(10, 6))
plt.scatter(n, e1, s=15, c='blue', edgecolors='none', marker='o')

plt.xlabel('n')
plt.ylabel('e1')
#plt.title(f'Energy Spectrum (m={m}, gamma={gamma}, lambda={lambda_val})')
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```
