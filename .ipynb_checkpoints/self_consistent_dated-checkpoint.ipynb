```python
#tight-binding d-wave BCS superconducting gap
import numpy as np

# --- 1. 参数设置 (必须与多变量版本完全一致) ---
t = 1.0
t1 = 0.0
t2 = 0.0
g = 2.0 * t
mu = 0.0

N = 32
k_points = np.linspace(-np.pi, np.pi, N, endpoint=False)
kx, ky = np.meshgrid(k_points, k_points)
Total_K_Points = N**2

# 形状因子 (d-wave 常见 cosx-cosy, 这里根据你之前的代码设为 1.0)
def gamma_func(kx, ky):
    gamma = np.cos(kx) - np.cos(ky)
    #gamma = 1.0
    return gamma

# 能带色散
def dispersion(kx, ky):
    ek = -2 * t * (np.cos(kx) + np.cos(ky))
    ek -= 4 * t1 * np.cos(kx) * np.cos(ky)
    ek -= 2 * t2 * (np.cos(2*kx) + np.cos(2*ky))
    ek -= mu
    return ek

xi_2d = dispersion(kx, ky)
g_2d = gamma_func(kx, ky)

# --- 2. 单变量自洽迭代求解 ---
def solve_bcs_scalar(initial_delta=0.5, tol=1e-9, max_iter=1000):
    delta = initial_delta
    
    for i in range(max_iter):
        # 计算当前的准粒子能量 Ek = sqrt(xi^2 + Delta_k^2)
        # Delta_k = delta * gamma_k
        delta_k = delta * g_2d
        Ek = np.sqrt(xi_2d**2 + delta_k**2)
        
        # 避免分母为零
        Ek = np.where(Ek == 0, 1e-12, Ek)
        
        # BCS Gap Equation 核心求和项:
        # delta = (g/N) * sum_k [ (gamma_k^2 * delta) / (2 * Ek) ]
        # 注意：这里的权重必须是 1/Total_K_Points
        new_delta = (g / Total_K_Points) * np.sum((g_2d**2 * delta) / (2 * Ek))
        
        # 检查收敛性
        if abs(new_delta - delta) < tol:
            print(f"迭代成功！在第 {i} 步收敛。")
            return new_delta
        
        delta = new_delta
        
    print("未能在最大迭代次数内收敛")
    return delta

# 执行计算
delta_final = solve_bcs_scalar()
print(f"单变量计算出的 Delta: {delta_final:.8f}")

# --- 3. 验证: 还原成 1024 个变量后的残差 ---
# 如果 1024 个变量的代码是正确的，那么代入下面这个 theta 应该让残差为 0
theta_k = 0.5 * np.arctan2(delta_final * g_2d, xi_2d)
# 计算残差向量的一个分量作为示例
residual_sample = 2 * xi_2d * np.sin(2*theta_k) - (g_2d * delta_final) * np.cos(2*theta_k)
print(f"带入单变量解后的最大残差: {np.max(np.abs(residual_sample)):.2e}")
```

```python
#s-wave
import numpy as np
from scipy.optimize import root

t = 1.0
t1 = 0.0
t2 = 0.0
g = 2.0*t
mu = 0.0

N = 32
k_points = np.linspace(-np.pi, np.pi, N, endpoint=False)
kx, ky = np.meshgrid(k_points, k_points)
Total_K_Points = N**2

def gamma_func(kx, ky):
    gamma = 1.0
    #gamma = np.cos(kx) - np.cos(ky)
    return gamma

def dispersion(kx, ky):
    ek = -2 * t * (np.cos(kx) + np.cos(ky))
    ek -= 4 * t1 * np.cos(kx) * np.cos(ky)
    ek -= 2 * t2 * (np.cos(2*kx) + np.cos(2*ky))
    ek -=mu
    return ek

xi_2d = dispersion(kx, ky)
g_2d = gamma_func(kx, ky)

def bcs_2d_system_with_jac(theta_flat):
    theta = theta_flat.reshape(xi_2d.shape)
    sin_2t = np.sin(2 * theta)
    cos_2t = np.cos(2 * theta)
    
    gap_sum = (g / Total_K_Points) * np.sum(g_2d * sin_2t)
    residual = 2 * xi_2d * sin_2t - g_2d * cos_2t * gap_sum
    
    vec_A = 2 * g_2d * cos_2t
    vec_B = (g / Total_K_Points) * g_2d * cos_2t 
    
    jac = -np.outer(vec_A.flatten(), vec_B.flatten())
    
    diag_elements = 4 * xi_2d.flatten() * cos_2t.flatten() + 2 * g_2d * sin_2t.flatten() * gap_sum
    
    np.fill_diagonal(jac, jac.diagonal() + diag_elements)
    
    return residual.flatten(), jac

delta_guess = 0.37
initial_theta = 0.5 * np.arctan2(delta_guess * g_2d, xi_2d)
initial_theta_flat = initial_theta.flatten()

sol = root(
    bcs_2d_system_with_jac, 
    initial_theta_flat, 
    method = 'lm', 
    tol=1e-6,
    jac=True
)

if sol.success:
    theta_res = sol.x.reshape(N, N)
    delta_final = g/2/Total_K_Points * np.sum(g_2d * np.sin(2 * theta_res))
    print(f" Delta: {delta_final:.6f}")
else:
    print("error", sol.message)
```

```python
#d-wave
import numpy as np
from scipy.optimize import root

t = 1.0
t1 = 0.0
t2 = 0.0
g = 2.0*t
mu = 0.0

N = 32
k_points = np.linspace(-np.pi, np.pi, N, endpoint=False)
kx, ky = np.meshgrid(k_points, k_points)
Total_K_Points = N**2

def gamma_func(kx, ky):
    gamma = 1.0
    gamma = np.cos(kx) - np.cos(ky)
    return gamma

def dispersion(kx, ky):
    ek = -2 * t * (np.cos(kx) + np.cos(ky))
    ek -= 4 * t1 * np.cos(kx) * np.cos(ky)
    ek -= 2 * t2 * (np.cos(2*kx) + np.cos(2*ky))
    ek -=mu
    return ek

xi_2d = dispersion(kx, ky)
g_2d = gamma_func(kx, ky)

def bcs_2d_system(theta_flat):
    theta = theta_flat.reshape(N, N)
    
    sin_2t = np.sin(2 * theta)
    cos_2t = np.cos(2 * theta)
    
    gap_sum = 0.5*g/Total_K_Points * np.sum(g_2d * sin_2t)
    
    residual = 2 * xi_2d * sin_2t - 2 * g_2d * cos_2t * gap_sum
    return residual.flatten()

delta_guess = 0.4
initial_theta = 0.5 * np.arctan2(delta_guess * g_2d, xi_2d)
initial_theta_flat = initial_theta.flatten()

sol = root(
    bcs_2d_system, 
    initial_theta_flat, 
    method='krylov', 
    tol=1e-6,
)

if sol.success:
    theta_res = sol.x.reshape(N, N)
    delta_final = g/2/Total_K_Points * np.sum(g_2d * np.sin(2 * theta_res))
    print(f" Delta: {delta_final:.6f}")
else:
    print("error", sol.message)
```
